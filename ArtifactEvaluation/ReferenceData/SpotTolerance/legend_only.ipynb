```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
```

# Legend Visualizations

This notebook contains standalone legend visualizations for the interruption handling analysis.

```python
# Legend for Line Plots (RPS/Latency time series)
# Matches the default matplotlib color cycle order as used in the plot functions
fig, ax = plt.subplots(figsize=(14, 1))
ax.axis('off')

# Colors match the order in which approaches are plotted:
# On-demand (gray), ShuntServe (blue), Concurrent Init (orange), Request Migration (green), No Handle (red), Avg (black dashed)
colors = ['#808080', '#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', 'black']
labels = ['On-demand Only', 'ShuntServe', 'Concurrent Initialization', 'Request Migration', 'No Handle', 'Average']
linestyles = ['-', '-', '-', '-', '-', '--']

# Create legend elements
legend_elements = [Line2D([0], [0], color=color, linewidth=3, linestyle=linestyle, label=label) 
                   for color, linestyle, label in zip(colors, linestyles, labels)]

# Create legend
legend = ax.legend(handles=legend_elements, loc='center', 
                   fontsize=18, ncol=6,
                   columnspacing=1.5, handlelength=2.5, edgecolor='black')

plt.tight_layout()
plt.savefig('preemption_tolerance_legend.pdf', dpi=300, bbox_inches='tight')
plt.show()
print("Saved: preemption_tolerance_legend.pdf")
```

```python
# Legend for Bar Plots (Throughput/Cost Analysis)
fig, ax = plt.subplots(figsize=(12, 1.5))
ax.axis('off')

# Define colors - these match the bar plot order
colors = ['#808080', '#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']  # Gray: On-demand, Blue: ShuntServe, Orange: Concurrent, Green: Request, Red: No Handle
labels = ['On-demand Only', 'ShuntServe', 'Concurrent Initialization', 'Request Migration', 'No Handle']

# Create legend elements (rectangles without patterns)
legend_elements = [mpatches.Rectangle((0, 0), 1, 1, 
                                      facecolor=color, 
                                      edgecolor='black', 
                                      linewidth=1.5, 
                                      alpha=0.7,
                                      label=label) 
                   for color, label in zip(colors, labels)]

# Create legend
legend = ax.legend(handles=legend_elements, loc='center', 
                   fontsize=18, framealpha=0.9, ncol=5,
                   columnspacing=1.5, handlelength=2.5)

plt.tight_layout()
plt.savefig('legend_bar_plot.pdf', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
print("Saved: legend_bar_plot.pdf")
```
